```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def show_comparision(original,robot,pervitt,canny):
  plt.figure(figsize=(12,10))
  plt.subplot(2,2,1)
  plt.imshow(original,cmap='gray')
  plt.title("Original grayscale image")
  plt.axis('off')

  plt.subplot(2,2,2)
  plt.imshow(robot,cmap='gray')
  plt.title("Robot edge detection")
  plt.axis('off')

  plt.subplot(2,2,3)
  plt.imshow(pervitt,cmap='gray')
  plt.title("Pervitt edge detection")
  plt.axis('off')

  plt.subplot(2,2,4)
  plt.imshow(canny,cmap='gray')
  plt.title("Canny edge detection")
  plt.axis('off')

  plt.show()
```

```python
img=cv2.imread('parrot.jpg',0)
if img is None:
  print("Image not found")
else:
  sobelx=cv2.Sobel(img, cv2.CV_64F,1,0,ksize=3)
  sobely=cv2.Sobel(img, cv2.CV_64F,0,1,ksize=3)
  sobel_combined=cv2.magnitude(sobelx,sobely)
  kernelx=np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
  kernely=np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
  perwittx=cv2.filter2D(img, cv2.CV_64F, kernelx)
  perwitty=cv2.filter2D(img, cv2.CV_64F, kernely)
  perwitt_combined=cv2.magnitude(perwittx,perwitty)
  canny=cv2.Canny(img,100,200)
  show_comparision(img,sobel_combined,perwitt_combined,canny)
#
```
